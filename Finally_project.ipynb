```python
# 8
import pandas as pd
from scipy import stats
df = pd.read_csv("AB_test.csv")
df.head()
```

```python
arpu = df.groupby(['experiment_num','experiment_group'])['revenue'].mean().reset_index()
print(arpu)
```

```python
results = {}
for exp in df['experiment_num'].unique():
    test_group = df[(df['experiment_num']==exp) & (df['experiment_group']=='test')]['revenue']
    control_group = df[(df['experiment_num']==exp) & (df['experiment_group']=='control')]['revenue']
    
    t_stat, p_val = stats.ttest_ind(test_group, control_group, equal_var=False)
    results[exp] = {'t_stat': t_stat, 'p_value': p_val}

results
```
